**1.2.3 Exact solution**

The two stationary solutions to the wave equations are real and imaginary components of the waves

$$
u(x, y, t) = e^{i(k_xx + k_yy - \omega t)} \tag{1}
$$

Show that this satisfies the wave equatiot: 

$$
\frac{\partial^{2}u}{\partial t^{2}}=c^{2}\,\nabla^{2}u , \qquad \nabla^{2}u=\frac{\partial^{2}u}{\partial x^{2}}+\frac{\partial^{2}u}{\partial y^{2}} . \tag{2}
$$

The solving for the Laplacien by differentiating twice with respect to both $x$ and $y$: 

$$\frac{\partial u}{\partial x} = ik_xe^{i(k_xx+k_yy -\omega t )}$$

so 

$$
\frac{\partial ^2 u}{\partial x^2} = (ik_x)^2 e^{i(k_xx+k_yy -\omega t )} = -k_x^2u
$$

And for $y$ :
$$
\frac{\partial ^2 u}{\partial y^2}= -k_y^2u
$$



So we can write the Laplacien as: 
$$
\nabla^{2}u = \frac{\partial^{2}u}{\partial x^{2}}+\frac{\partial^{2}u}{\partial y^{2}} = -(k_{x}^{2}+k_{y}^{2})\,u . \tag{3}
$$


Inserting (3) and (4) into the (2) we get: 

$$
-\omega^2 u = c^2(-(k_{x}^{2}+k_{y}^{2})u) \rightarrow -\omega^2 = c^2(-(k_{x}^{2}+k_{y}^{2}))
$$

Wich gives us finally: 

$$
\omega^2 = c^2((k_{x}^{2}+k_{y}^{2})) \tag{4}
$$


1.2.4 Dispersion coefficient#

Assume that $m_x = m_y$ such that $k_x = k_y$. A discrete version of eq (1) will then be:

$$
u_{ij}^n = e^{i(kh(i + j) - \tilde{\omega}\Delta t)}\tag{6}
$$


Where $\tilde{\omega}$ is the numerical dispersion coefficient ,i.e the numerical approximation of the exact $\omega$. Insert (6) into the discretized eqution (7) and show that for a CFL of $C= \frac{1}{\sqrt{2}}$ we get $\tilde{\omega} = \omega $. 


$$
\frac{u_{ij}^{n+1 } + 2 u_{ij}^n + u_{ij}^{n-1}}{\Delta t^2}  = c^2 (\frac{u_{i+1j}^{n} - 2 u_{ij}^{n} + u_{i-1j}^{n}}{h^2} + \frac{u_{ij+1}^{n} - 2 u_{ij}^{n} + u_{ij-1}^{n}}{h^2})\tag{7}
$$

$ C = \frac{c \Delta t }{\Delta x}$  and $\omega = c \sqrt{k_x^2 + k_y^2}$

Where $x_i = i h, y_j = jh, \Delta t \equiv \Delta t$

So the Temporal Part from (6) is then:

$u_{ij}^{n+1} = e^{i(kh(i + j) - \tilde{\omega}(n+1)\Delta t)}= e^{-i \tilde{\omega}\Delta t} u_{ij}^n $

and 

$u_{ij}^{n-1} = e^{i(kh(i + j) - \tilde{\omega}(n-1)\Delta t)}= e^{i \tilde{\omega}\Delta t} u_{ij}^n $



So inserting (6) into the discretized equation (7) we get a LHS of (7):

$$
\frac{e^{-i \tilde{\omega}\Delta t} u_{ij}^n  + 2 u_{ij}^n + e^{i \tilde{\omega}\Delta t} u_{ij}^n }{\Delta t^2}  \rightarrow \frac{e^{-i \tilde{\omega}\Delta t} + 2 + e^{i \tilde{\omega}\Delta t}}{\Delta t^2}  u_{ij}^n
$$

With the trig identities: 

$$
e^{\pm i\theta} = cos \theta \pm i sin \theta 
$$
And
$$
cos \theta -1 = -2 sin^2\frac{\theta}{2}
$$

This simplifies to: 

$$
e^{-\tilde{\omega}\Delta t} + e^{\tilde{\omega}\Delta t} -2 = 2 cos(\tilde{\omega}\Delta t) - 2 = -4sin^2(\frac{\tilde{\omega}\Delta t}{2})
$$
$$\downarrow$$
$$
 \frac{u_{i,j}^{,n+1}-2u_{i,j}^{,n}+u_{i,j}^{,n-1}}{\Delta t^{2}} = \frac{-4sin^2(\frac{\tilde{\omega}\Delta t}{2})}{\Delta t^ 2}\tag{8}
$$

$$
c^2 ( ...) = -c^2 \frac{8sin^2\frac{kh}{2}}{h^2} u_{i,j}^n
$$

Cancelling the common factor $u_{i,j}^n$ and using $C = \frac{c\Delta t}{h}$ then gives:

$$
sin^2\frac{\tilde{\omega}\Delta t}{2} = 2C^2sin^2(\frac{kh}{2})
$$

and for $C = \frac{1}{\sqrt{2}}$ we get $$\tilde{\omega}= \omega$$
